```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
plt.style.use('seaborn-v0_8-whitegrid')
plt.rcParams['figure.figsize'] = (10, 6)
plt.rcParams['font.size'] = 11
Path('figures').mkdir(exist_ok=True)
```

```python
df = pd.read_excel('sales.xlsx', sheet_name='Sales')
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
```

```python
print(df.head()); print(df.shape); print(df.columns.tolist())
df.info()
print(df.describe(include='all'))
print(df.isna().sum())
print(df.duplicated().sum())
print(df['Category'].value_counts())
print(df[['Sales','Profit']].agg(['min','max']))
```

```python
required = ['Date','Region','Category','Sales','Profit']
df = df.dropna(subset=required).copy()
df = df.drop_duplicates()
df['Region'] = df['Region'].str.strip()
df['Category'] = df['Category'].str.strip()
df['Sales'] = pd.to_numeric(df['Sales'], errors='coerce')
df['Profit'] = pd.to_numeric(df['Profit'], errors='coerce')
df = df.dropna(subset=['Sales','Profit'])
print(df.shape)
```

```python
region_summary = (df.groupby('Region', as_index=False)
    .agg(Sales=('Sales','sum'), Profit=('Profit','sum'), Orders=('Product','count'))
    .sort_values('Sales', ascending=False))
print(region_summary)
```

```python
df['Month'] = df['Date'].dt.to_period('M').dt.to_timestamp()
monthly = (df.groupby('Month', as_index=False)
    .agg(Sales=('Sales','sum'), Profit=('Profit','sum'))
    .sort_values('Month'))
print(monthly)
```

```python
category_region = (df.groupby(['Region','Category'], as_index=False)
    .agg(Sales=('Sales','sum'), AvgProfit=('Profit','mean')))
print(category_region)
```

```python

fig, ax = plt.subplots(figsize=(10, 6))
ax.bar(region_summary['Region'], region_summary['Sales'], color='#5B9BD5', edgecolor='white')
ax.set_title('Өңірлер бойынша сатылым', fontsize=15, fontweight='bold')
ax.set_xlabel('Өңір'); ax.set_ylabel('Сатылым, теңге')
ax.tick_params(axis='x', rotation=30); ax.grid(axis='y', alpha=0.25)
fig.tight_layout()
fig.savefig('figures/01_sales_by_region.png', dpi=300, bbox_inches='tight', facecolor='white')
plt.show()
```

```python

fig, ax = plt.subplots(figsize=(11, 6))
ax.plot(monthly['Month'], monthly['Sales'], marker='o', linewidth=2.2, color='#1F4E78', label='Сатылым')
ax.set_title('Айлар бойынша сатылым динамикасы', fontsize=15, fontweight='bold')
ax.set_xlabel('Ай'); ax.set_ylabel('Сатылым, теңге')
ax.legend(title='Көрсеткіш'); ax.grid(alpha=0.25)
fig.autofmt_xdate(); fig.tight_layout()
fig.savefig('figures/02_monthly_sales.png', dpi=300, bbox_inches='tight', facecolor='white')
plt.show()
```

```python

fig, ax = plt.subplots(figsize=(9, 6))
for category, part in df.groupby('Category'):
    ax.scatter(part['Sales'], part['Profit'], label=category, alpha=0.65, s=45)
ax.axhline(0, color='red', linewidth=1, linestyle='--')
ax.set_title('Сатылым мен пайда арасындағы байланыс', fontsize=15, fontweight='bold')
ax.set_xlabel('Сатылым, теңге'); ax.set_ylabel('Пайда, теңге')
ax.legend(title='Санат', bbox_to_anchor=(1.02, 1), loc='upper left')
fig.tight_layout()
fig.savefig('figures/03_sales_profit_scatter.png', dpi=300, bbox_inches='tight', facecolor='white')
plt.show()
```

```python

fig, ax = plt.subplots(figsize=(9, 6))
ax.hist(df['Sales'].dropna(), bins=20, color='#70AD47', edgecolor='white', alpha=0.85)
ax.set_title('Транзакция сатылымының үлестірімі', fontsize=15, fontweight='bold')
ax.set_xlabel('Сатылым, теңге'); ax.set_ylabel('Жиілік')
ax.grid(axis='y', alpha=0.25)
fig.tight_layout()
fig.savefig('figures/04_sales_histogram.png', dpi=300, bbox_inches='tight', facecolor='white')
plt.show()
```
