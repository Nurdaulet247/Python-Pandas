```python
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.ensemble import IsolationForest
from sklearn.neighbors import LocalOutlierFactor
from sklearn.preprocessing import StandardScaler

plt.style.use("seaborn-v0_8-whitegrid")
pd.set_option("display.max_columns", None)

RANDOM_STATE = 42
```

```python
url = "https://raw.githubusercontent.com/numenta/NAB/master/data/realKnownCause/machine_temperature_system_failure.csv"

df = pd.read_csv(url)

print("Деректер көлемі:", df.shape)

display(df.head())
```

```python
df.info()
print(df.isnull().sum())
print("", df.duplicated().sum())
display(df["value"].describe())
```

```python
df["timestamp"] = pd.to_datetime(df["timestamp"])

df = df.sort_values("timestamp")
df = df.drop_duplicates(subset="timestamp")
df = df.reset_index(drop=True)

print("", df["timestamp"].min())
print("", df["timestamp"].max())

print("", df.shape)
```

```python
plt.figure(figsize=(14, 5))

plt.subplot(1, 2, 1)
sns.histplot(df["value"], bins=50, kde=True)
plt.title("Температураның үлестірімі")
plt.xlabel("Температура")

plt.subplot(1, 2, 2)
sns.boxplot(x=df["value"])
plt.title("Температураның Boxplot диаграммасы")
plt.xlabel("Температура")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(15, 5))

plt.plot(df["timestamp"], df["value"], linewidth=0.7)

plt.title("Машина температурасының уақыт бойынша өзгеруі")
plt.xlabel("Уақыт")
plt.ylabel("Температура")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
Q1 = df["value"].quantile(0.25)
Q3 = df["value"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df["iqr_anomaly"] = (
    (df["value"] < lower) |
    (df["value"] > upper)
)

print("Төменгі шекара:", round(lower, 3))
print("Жоғарғы шекара:", round(upper, 3))

print("IQR бойынша аномалия саны:", df["iqr_anomaly"].sum())
```

```python
WINDOW = 24

df["diff_1"] = df["value"].diff()

df["rolling_mean"] = df["value"].rolling(WINDOW).mean()

df["rolling_std"] = df["value"].rolling(WINDOW).std()

df["deviation"] = df["value"] - df["rolling_mean"]

display(df.head(10))
```

```python
feature_cols = [
    "value",
    "diff_1",
    "rolling_mean",
    "rolling_std",
    "deviation"
]

model_df = df.dropna(subset=feature_cols).copy()

print("", model_df.shape)
```

```python
time_step = model_df["timestamp"].diff().median()

print("Өлшемдер арасындағы уақыт:", time_step)
print("24 өлшемнің ұзақтығы:", time_step * WINDOW)
```

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(model_df[feature_cols])

print("Масштабталған деректер көлемі:", X_scaled.shape)
```

```python
iso = IsolationForest(
    n_estimators=300,
    contamination=0.01,
    random_state=42
)

model_df["iso_label"] = iso.fit_predict(X_scaled)

model_df["iso_anomaly"] = model_df["iso_label"] == -1

model_df["iso_score"] = -iso.score_samples(X_scaled)

print("Isolation Forest аномалия саны:",
      model_df["iso_anomaly"].sum())

print("Аномалия үлесі:",
      round(model_df["iso_anomaly"].mean() * 100, 2), "%")
```

```python
lof = LocalOutlierFactor(
    n_neighbors=35,
    contamination=0.01
)

model_df["lof_label"] = lof.fit_predict(X_scaled)

model_df["lof_anomaly"] = model_df["lof_label"] == -1

model_df["lof_score"] = -lof.negative_outlier_factor_

print("LOF аномалия саны:",
      model_df["lof_anomaly"].sum())

print("Аномалия үлесі:",
      round(model_df["lof_anomaly"].mean() * 100, 2), "%")
```

```python
model_df["both_anomaly"] = (
    model_df["iso_anomaly"] &
    model_df["lof_anomaly"]
)

model_df["any_anomaly"] = (
    model_df["iso_anomaly"] |
    model_df["lof_anomaly"]
)

comparison = pd.DataFrame({
    "Әдіс": [
        "Isolation Forest",
        "LOF",
        "Екі модель бірдей анықтады",
        "Кемінде бір модель анықтады"
    ],

    "Аномалия саны": [
        model_df["iso_anomaly"].sum(),
        model_df["lof_anomaly"].sum(),
        model_df["both_anomaly"].sum(),
        model_df["any_anomaly"].sum()
    ]
})

comparison["Үлесі (%)"] = (
    comparison["Аномалия саны"] /
    len(model_df) * 100
)

display(comparison.round(2))
```

```python
cross = pd.crosstab(
    model_df["iso_anomaly"],
    model_df["lof_anomaly"],
    rownames=["Isolation Forest"],
    colnames=["LOF"]
)

display(cross)
```

```python
plt.figure(figsize=(16, 6))

plt.plot(
    model_df["timestamp"],
    model_df["value"],
    color="steelblue",
    linewidth=0.7,
    label="Қалыпты деректер"
)

plt.scatter(
    model_df.loc[model_df["iso_anomaly"], "timestamp"],
    model_df.loc[model_df["iso_anomaly"], "value"],
    color="red",
    s=20,
    label="Isolation Forest"
)

plt.scatter(
    model_df.loc[model_df["lof_anomaly"], "timestamp"],
    model_df.loc[model_df["lof_anomaly"], "value"],
    color="orange",
    s=20,
    label="LOF"
)

plt.title("Екі алгоритм бойынша анықталған аномалиялар")
plt.xlabel("Уақыт")
plt.ylabel("Температура")
plt.legend()

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=model_df,
    x="value",
    y="diff_1",
    hue="both_anomaly",
    palette={
        False: "steelblue",
        True: "red"
    },
    s=25
)

plt.title("Температура және оның өзгерісі")
plt.xlabel("Температура")
plt.ylabel("Алдыңғы өлшеммен салыстырғандағы өзгеріс")

plt.legend(title="Екі модель анықтады")
plt.show()
```

```python
columns = [
    "timestamp",
    "value",
    "diff_1",
    "rolling_mean",
    "rolling_std",
    "iso_score",
    "lof_score",
    "iso_anomaly",
    "lof_anomaly"
]

top_anomalies = model_df.nlargest(
    10,
    "iso_score"
)[columns]

display(top_anomalies.round(3))
```

```python
results = []

for contamination in [0.005, 0.01, 0.02]:

    model = IsolationForest(
        n_estimators=200,
        contamination=contamination,
        random_state=42
    )

    labels = model.fit_predict(X_scaled)

    results.append({
        "Contamination": contamination,
        "Аномалия саны": (labels == -1).sum()
    })

display(pd.DataFrame(results))
```

```python
results_lof = []

for k in [15, 35, 70]:

    model = LocalOutlierFactor(
        n_neighbors=k,
        contamination=0.01
    )

    labels = model.fit_predict(X_scaled)

    results_lof.append({
        "Көршілер саны": k,
        "Аномалия саны": (labels == -1).sum()
    })

display(pd.DataFrame(results_lof))
```
